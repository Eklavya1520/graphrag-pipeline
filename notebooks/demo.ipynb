# Agentic GraphRAG Pipeline — End-to-End Demo

This notebook walks through the full pipeline:
1. **Ingestion** — Agent A extracts entities and relationships from text and loads them into Neo4j
2. **Query rewriting** — Agent B cleans and structures a vague user query
3. **Graph retrieval** — Cypher traversal fetches structured context from Neo4j
4. **Vector fallback** — FAISS kicks in if graph context is thin
5. **Self-correcting generation** — LLM generates an answer, critic scores it, retries if needed
6. **Ragas evaluation** — Measures faithfulness, relevancy, recall, and precision

> **Before running**: copy `.env.example` to `.env` and fill in your API keys and Neo4j credentials.

## Step 2 — Agent A: Ingest a Document

Agent A reads raw text, extracts `(entity, relation, entity)` triples via an LLM with structured output, and loads them into Neo4j.

````python
from agents.graph_builder import GraphBuilder
from graph.neo4j_client import Neo4jClient

sample_text = """
The Transformer model was introduced by Vaswani et al. in 2017 in the paper 'Attention Is All You Need'.
It uses a self-attention mechanism to compute relationships between all tokens in a sequence simultaneously.
Multi-head attention allows the model to attend to different representation subspaces in parallel.
BERT, developed by Google, uses only the encoder stack of the Transformer.
BERT is pre-trained using masked language modeling and next sentence prediction.
GPT, developed by OpenAI, uses only the decoder stack and performs autoregressive language modeling.
"""

with Neo4jClient() as client:
    builder = GraphBuilder(client)
    stats = builder.ingest_text(sample_text, source_label='notebook_demo')

print(f"Ingestion complete:")
print(f"  Entities written  : {stats['entities']}")
print(f"  Relations written : {stats['relations']}")
````

````python
# Quick peek at what was ingested
from graph.neo4j_client import Neo4jClient

with Neo4jClient() as client:
    entities = client.query(
        'MATCH (n:Entity) RETURN n.name AS name, n.type AS type, n.description AS desc LIMIT 10'
    )
    relations = client.query(
        'MATCH (a:Entity)-[r:RELATION]->(b:Entity) RETURN a.name AS src, r.type AS rel, b.name AS tgt LIMIT 10'
    )

print('=== Entities ===')
for e in entities:
    print(f"  [{e['type']}] {e['name']}: {e['desc'][:60]}..." if e['desc'] and len(e['desc']) > 60 else f"  [{e['type']}] {e['name']}: {e['desc']}")

print()
print('=== Relationships ===')
for r in relations:
    print(f"  {r['src']} --[{r['rel']}]--> {r['tgt']}")
````

## Step 3 — Agent B: Query Rewriting

Agent B takes a vague query and rewrites it into a specific, structured question. It also extracts entity keywords used to seed the Cypher graph traversal.

````python
from agents.query_rewriter import QueryRewriter

rewriter = QueryRewriter()

test_queries = [
    'tell me about transformer',
    'how does bert work',
    'difference between rnn and transformer',
]

for q in test_queries:
    result = rewriter.rewrite(q)
    print(f'Original : {q}')
    print(f'Rewritten: {result.rewritten_query}')
    print(f'Keywords : {result.entity_keywords}')
    print(f'Type     : {result.query_type}')
    print()
````

## Step 4 — Graph Retrieval

The graph retriever runs Cypher queries against Neo4j using the keywords extracted by Agent B.

````python
from graph.neo4j_client import Neo4jClient
from graph.graph_retriever import GraphRetriever

keywords = ['Transformer', 'attention', 'BERT']

with Neo4jClient() as client:
    retriever = GraphRetriever(client)
    context = retriever.retrieve(keywords, top_k=8)

print(f'Retrieved {len(context)} context items from the graph:')
print()
for i, item in enumerate(context, 1):
    print(f'  {i}. {item}')
````

## Step 5 — Full Pipeline (LangGraph)

Now we run the complete self-correcting pipeline. LangGraph orchestrates the full flow:

```
rewrite_query → retrieve_from_graph → [vector fallback?] → generate_answer → score_confidence → [retry?]
```

````python
from pipeline.graph import pipeline
from pipeline.state import initial_state
import time

question = 'What is the attention mechanism and how does BERT use it?'

print(f'Question: {question}\n')
print('Running pipeline...')

start = time.time()
result = pipeline.invoke(initial_state(question))
elapsed = time.time() - start

print(f'Done in {elapsed:.1f}s\n')
print('=' * 60)
print('ANSWER')
print('=' * 60)
print(result['answer'])
print()
print(f"Confidence  : {result['confidence']:.2f}")
print(f"Iterations  : {result['iterations']}")
print(f"Rewritten Q : {result['rewritten_query']}")
print(f"Graph ctx   : {len(result['graph_context'])} items")
print(f"Vector ctx  : {len(result['vector_context'])} items")
````

### Try a few more questions

````python
questions = [
    'How does GPT differ from BERT?',
    'What did Vaswani et al. propose?',
    'Explain masked language modeling',
]

for q in questions:
    state = pipeline.invoke(initial_state(q))
    conf  = state['confidence']
    iters = state['iterations']
    ans   = state['answer'][:200] + '...' if len(state['answer']) > 200 else state['answer']
    low   = ' [LOW CONF]' if state.get('low_confidence_warning') else ''
    print(f'Q: {q}')
    print(f'A: {ans}')
    print(f'   conf={conf:.2f}  iters={iters}{low}')
    print()
````

## Step 6 — Ragas Evaluation

We evaluate the pipeline on a small test set using [Ragas](https://docs.ragas.io/). Ragas uses an LLM internally to compute faithfulness and answer relevancy scores.

> **Cost note**: each question costs approximately \$0.003–0.005 in LLM API calls. Running the full 15-question test set costs under \$0.10.

````python
import json
from evaluation.ragas_eval import RagasEvaluator
from pipeline.graph import pipeline
from pipeline.state import initial_state

# Load evaluation questions
with open('../data/eval_questions.json') as f:
    eval_data = json.load(f)

# Use first 5 questions for this demo (full eval in scripts/evaluate.py)
eval_data = eval_data[:5]

print(f'Evaluating on {len(eval_data)} questions...')

samples = []
for item in eval_data:
    state    = pipeline.invoke(initial_state(item['question']))
    contexts = state.get('graph_context', []) + state.get('vector_context', [])
    samples.append({
        'question'    : item['question'],
        'answer'      : state.get('answer', ''),
        'contexts'    : contexts or ['No context retrieved'],
        'ground_truth': item['ground_truth'],
    })

evaluator = RagasEvaluator()
scores = evaluator.evaluate(samples)
evaluator.print_summary(scores)
````

## Visualise the Knowledge Graph

Quick subgraph visualisation using matplotlib. For a proper interactive view, open Neo4j Browser at `http://localhost:7474` and run:
```cypher
MATCH (n:Entity)-[r:RELATION]->(m:Entity) RETURN n, r, m LIMIT 50
```

````python
import networkx as nx
import matplotlib.pyplot as plt
from graph.neo4j_client import Neo4jClient

with Neo4jClient() as client:
    edges = client.query(
        'MATCH (a:Entity)-[r:RELATION]->(b:Entity) RETURN a.name AS src, r.type AS rel, b.name AS tgt LIMIT 30'
    )

G = nx.DiGraph()
for e in edges:
    G.add_edge(e['src'], e['tgt'], label=e['rel'])

plt.figure(figsize=(14, 9))
pos = nx.spring_layout(G, seed=42, k=2.5)
nx.draw_networkx_nodes(G, pos, node_size=1800, node_color='#4C9BE8', alpha=0.85)
nx.draw_networkx_labels(G, pos, font_size=8, font_color='white', font_weight='bold')
nx.draw_networkx_edges(G, pos, edge_color='#888', arrows=True, arrowsize=20, width=1.5)
edge_labels = nx.get_edge_attributes(G, 'label')
nx.draw_networkx_edge_labels(G, pos, edge_labels=edge_labels, font_size=7, font_color='#333')
plt.title('Knowledge Graph — Neo4j Subgraph (top 30 relationships)', fontsize=13, pad=15)
plt.axis('off')
plt.tight_layout()
plt.show()
print(f'Graph has {G.number_of_nodes()} nodes and {G.number_of_edges()} edges in this view.')
````

## Summary

| Component | What it does |
|---|---|
| **Agent A (GraphBuilder)** | Parses raw text, extracts triples, loads Neo4j |
| **Agent B (QueryRewriter)** | Normalises vague queries, extracts Cypher-seed keywords |
| **GraphRetriever** | Runs Cypher neighbourhood queries against Neo4j |
| **VectorStore** | FAISS fallback when graph context is sparse |
| **AnswerGenerator** | Synthesises grounded answers from combined context |
| **Self-correction loop** | Retries up to 3x if confidence < 0.6 |
| **Ragas** | Evaluates faithfulness, relevancy, recall, precision |
